# Model comparison
Reads real test metrics generated by training runs.

```python
from pathlib import Path
import json, pandas as pd
rows=[]
for p in Path('../outputs').glob('*/test_metrics.json'):
    d=json.loads(p.read_text())
    rows.append({'model':p.parent.name,'accuracy':d['accuracy'],'macro_f1':d['macro_f1'],'weighted_f1':d['weighted_f1']})
display(pd.DataFrame(rows) if rows else 'Train models to generate measured metrics.')
```
